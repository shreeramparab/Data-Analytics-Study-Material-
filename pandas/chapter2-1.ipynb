'''Yes, in **Pandas** there are several functions similar to **`apply()`** that let you apply functions to data.

Here are the most common ones:

---

## 1️⃣ `apply()`

Applies a function to **rows or columns** of a DataFrame.

```python
df.apply(lambda x: x.max())
```

Example (row-wise):

```python
df.apply(lambda x: x['Salary'] / x['Years'], axis=1)
```

---

## 2️⃣ `map()`

Works on a **single Series (one column)**.

```python
df['Salary'].map(lambda x: x * 1.1)
```

Example:

```python
df['Gender'].map({'M':'Male','F':'Female'})
```

✔ Faster than `apply` for **single column transformations**.

---

## 3️⃣ `applymap()`

Applies a function to **every element of a DataFrame**.

```python
df.applymap(lambda x: str(x).upper())
```

Example:

```python
df.applymap(lambda x: x*2)
```

---

## 4️⃣ `transform()`

Used with **groupby** to transform values while keeping the **same shape**.

Example:

```python
df.groupby('Department')['Salary'].transform('mean')
```

---

## 5️⃣ `agg()` / `aggregate()`

Used for **aggregation operations**.

```python
df.groupby('Department')['Salary'].agg(['mean','max','min'])
```

---

## Quick comparison

| Function      | Works On           | Purpose                      |
| ------------- | ------------------ | ---------------------------- |
| `apply()`     | DataFrame / Series | Apply function row or column |
| `map()`       | Series only        | Transform values             |
| `applymap()`  | Entire DataFrame   | Element-wise operation       |
| `transform()` | GroupBy            | Transform but keep same size |
| `agg()`       | GroupBy / Series   | Aggregation                  |

---

💡 **Simple rule to remember**

```
map → single column
apply → row/column
applymap → entire dataframe
agg → aggregation
transform → groupby transformation
'''
