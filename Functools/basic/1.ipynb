### 🔹 1. map()

* Works on **Series (single column)**
* Applies function **element-wise**

```python
df['grade'] = df['marks'].map(lambda x: 'Pass' if x >= 40 else 'Fail')
```

---

### 🔹 2. apply()

* Works on **Series and DataFrame**
* Can be **row-wise or column-wise**

```python
# Column-wise
df['square'] = df['marks'].apply(lambda x: x**2)

# Row-wise
df['total'] = df.apply(lambda row: row['a'] + row['b'], axis=1)
```

---

### 🔹 3. assign()

* Used to **create new columns**
* Returns a **new DataFrame**

```python
df = df.assign(
    bonus = df['salary'] * 0.1,
    total_salary = lambda x: x['salary'] + x['bonus']
)
```

---

### 🔹 4. lambda

* Small **anonymous function**
* Syntax:

```python
lambda arguments: expression
```

Example:

```python
lambda x: x + 10
```

* Used inside `map()`, `apply()`, `assign()`

---

### 🔥 Quick Summary

* `map()` → element-wise (Series only)
* `apply()` → flexible (row/column)
* `assign()` → add new columns
* `lambda` → one-line function
